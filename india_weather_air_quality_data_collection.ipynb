```python
all_air_data = []

pollutant_columns = [
    "pm2_5",
    "pm10",
    "carbon_monoxide",
    "nitrogen_dioxide",
    "sulphur_dioxide",
    "ozone"
]

for index, row in city_coord.iterrows():

    city = row["city"]
    state_ut = row["state_ut"]
    latitude = row["latitude"]
    longitude = row["longitude"]

    air_params = {
        "latitude": latitude,
        "longitude": longitude,
        "start_date": "2026-07-09",
        "end_date": "2026-10-08",
        "hourly": pollutant_columns,
        "timezone": "Asia/Kolkata"
    }

    # Request data for the current city
    air_response = requests.get(air_url, params=air_params)
    air_response.raise_for_status()
    air_data = air_response.json()

    # Convert hourly API data into a DataFrame
    air_hourly_df = pd.DataFrame(air_data["hourly"])

    # Extract the date from each timestamp
    air_hourly_df["date"] = pd.to_datetime(
        air_hourly_df["time"]
    ).dt.date

    # Calculate daily averages
    air_daily_df = (
        air_hourly_df
        .groupby("date")[pollutant_columns]
        .mean()
        .reset_index()
    )

    # Label the records with their location
    air_daily_df["state_ut"] = state_ut
    air_daily_df["city"] = city

    # Save this city's daily records
    all_air_data.append(air_daily_df)

    print(f"Collected {city}: {len(air_daily_df)} records")

# Outside the loop: combine all city DataFrames
final_air_df = pd.concat(all_air_data, ignore_index=True)

print(final_air_df.head())
print(final_air_df.shape)
```
